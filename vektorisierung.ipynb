# Vektorisierung der Features

### Auswahl Features

```python
import nltk
from nltk.text import Text
from nltk.probability import FreqDist

with open("korpus\\goethe-faust1.txt", encoding="utf8") as input:
    text = input.read()

input.close()
nltk_text = Text(nltk.word_tokenize(text.lower())) #lower, damit es keine Doppelungen beim zählen später gibt
print("Länge:",len(nltk_text))
print("NLTK:",nltk_text)
```

### Features Numerisieren

```python
nltk_text_words = [t for t in nltk_text if t.isalpha()]  # Entferne Satzzeichen/Zahlen
number_words = len(nltk_text_words)
print("Wörter im Dokument:",number_words)

sentences = nltk.sent_tokenize(text) #start from the 1000th char and tokenize in sentences
number_sentences = len(sentences)
print("Sätze im Dokument:",number_sentences) #number of sentences

avg_sen_len = round(len(nltk_text_words) / len(sentences),2) #average sentence length
#print(avg_sen_len)
avg_word_len = round(sum(len(word) for word in nltk_text_words) / len(nltk_text_words),2)
#print(avg_word_len)
```

### Skalieren

```python
from sklearn.preprocessing import StandardScaler
from sklearn.metrics.pairwise import cosine_distances, cosine_similarity
import numpy as np

text1 = np.array([[number_words, number_sentences, avg_sen_len, avg_word_len]])
text1
```

```python
text2 = np.array([[133739, 13932, 4.58, 2.8]])
text2
```

```python
scaler = StandardScaler() #Werte Normalisiert
X_scaled = scaler.fit_transform([text1[0], text2[0]])
print("X_scale:\n",X_scaled)
print("")
print(cosine_distances([X_scaled[0]], [X_scaled[1]]))
```

```python
print("Cosine Distance:",cosine_distances(text1, text2))
print("Cosine Similarity:",cosine_similarity(text1, text2))
```
