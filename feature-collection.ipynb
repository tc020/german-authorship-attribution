# Surface Features

```python
import nltk
from nltk.text import Text
from nltk.probability import FreqDist
```

```python
with open("korpus\\goethe-faust1.txt", encoding="utf8") as input:
    text = input.read()

input.close()
nltk_text = Text(nltk.word_tokenize(text.lower())) #lower, damit es keine Doppelungen beim zählen später gibt
print("Länge:",len(nltk_text))
print("NLTK:",nltk_text)
```

# Number of Words

```python
nltk_text_words = [t for t in nltk_text if t.isalpha()]  # Entferne Satzzeichen/Zahlen
print("Es befinden sich",len(nltk_text_words),"Wörter im Dokument.")
```

# Number of Sentences

```python
sentences = nltk.sent_tokenize(text) #start from the 1000th char and tokenize in sentences
print("Es befinden sich",len(sentences),"Sätze im Dokument.\n") #number of sentences
print("Der 25. Satz ist:\n\n"+sentences[25])
```

# Average length of sentences

```python
avg_sen_len = len(nltk_text_words) / len(sentences) #average sentence length
avg_sen_len
```

# Average length of words

```python
avg_word_len = sum(len(word) for word in nltk_text_words) / len(nltk_text_words)
avg_word_len
```

# Count punctuations

```python
excl = 0
quest = 0
pct = 0
com = 0
semic = 0
colo = 0

for i in nltk_text:
    if i == "!":
        excl += 1
    if i == "?":
        quest += 1
    if i == ".":
        pct += 1
    if i == ",":
        com += 1
    if i == ";":
        semic += 1
    if i == ":":
        colo += 1

print("! ",excl)
print("? ",quest)
print(". ",pct)
print(", ",com)
print("; ",semic)
print(": ",colo)

print("\nSumme:",excl+quest+pct+com+semic+colo)
```
